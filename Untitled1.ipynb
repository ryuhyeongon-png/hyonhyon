```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://news.ycombinator.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

titles = soup.find_all("span", class_="titleline")

with open("news.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["タイトル"])
    for title in titles[:10]:
        writer.writerow([title.text])

print("保存完了！")
```

```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://news.ycombinator.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

titles = soup.find_all("span", class_="titleline")

with open("news_with_url.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["タイトル", "URL"])
    for title in titles[:10]:
        link = title.find("a")
        writer.writerow([link.text, link["href"]])

print("保存完了！")
```

```python
import requests
from bs4 import BeautifulSoup
import csv
from datetime import datetime

url = "https://realpython.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

articles = soup.find_all("h2")
除外 = ["Podcasts", "Explore Real Python", "New Releases"]

今日の日付 = datetime.now().strftime("%Y-%m-%d")
ファイル名 = f"articles_{今日の日付}.csv"

with open(ファイル名, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["日付", "タイトル"])
    for article in articles[:10]:
        text = article.text.strip()
        if text and text not in 除外:
            writer.writerow([今日の日付, text])

print(f"{ファイル名} に保存完了！")
```

```python
import pandas as pd
import glob

# すべてのCSVファイルを読み込んでまとめる
files = glob.glob("articles_*.csv")
df_all = pd.concat([pd.read_csv(f) for f in files])
print(df_all)

```

```python
import matplotlib.pyplot as plt

df_all['length'] = df_all['タイトル'].apply(len)
df_all.plot(kind='bar', x='タイトル', y='length', figsize=(12,5))
plt.title("Article Title Length")
plt.tight_layout()
plt.show()
```
