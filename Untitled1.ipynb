```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://news.ycombinator.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

titles = soup.find_all("span", class_="titleline")

with open("news.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["タイトル"])
    for title in titles[:10]:
        writer.writerow([title.text])

print("保存完了！")
```

```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://news.ycombinator.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

titles = soup.find_all("span", class_="titleline")

with open("news_with_url.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["タイトル", "URL"])
    for title in titles[:10]:
        link = title.find("a")
        writer.writerow([link.text, link["href"]])

print("保存完了！")
```

```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://realpython.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

articles = soup.find_all("h2")

with open("python_articles.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["タイトル"])
    for article in articles[:10]:
        text = article.text.strip()
        if text:
            writer.writerow([text])

print("保存完了！")
```

```python
除外 = ["Podcasts", "Explore Real Python", "New Releases"]

with open("python_articles_clean.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["タイトル"])
    for article in articles[:10]:
        text = article.text.strip()
        if text and text not in 除外:
            writer.writerow([text])

print("保存完了！")
```
